## Matplotlib example workflow

```python
# 0. import matplotlib and get it ready for plotting in Jupyter
import matplotlib.pyplot as plt

# 1. Prepare data
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2. Setup plot
fig, ax = plt.subplots(figsize = (10, 10))

# 3. Plot data
ax.plot(x, y)

# 4. Customise Plot
ax.set(title = "Simple Plot", xlabel = "x-axis", ylabel = "y-axis")

# 5. Save & show (you save the whole figure)
fig.savefig("/content/sample_data")
```

## Bar Plot

```python
# Make a plot from dictionary
nut_butter_prices = {"Almond butter": 10, "Peanut butter": 8, "Cashew butter": 12}

fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax.set(title = "Butter Store", xlabel = "", ylabel = "Pricess")
plt.show()
```

### Horizontal Bar Plot

```python

fig, ax = plt.subplots()
ax.barh(list(nut_butter_prices.keys()), list(nut_butter_prices.values()))
ax.set(title = "Butter store", xlabel = "Prices", ylabel = "")
plt.show()
```

# Two options for subplots

## Using Subplot with tuple format

```python
# Subplot option 1 (using tuple format)
np.random.seed(0)
# pass axis in tuple format
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows = 2, ncols = 2, figsize = (10, 5))

# Plot to each different axis
ax1.plot(x, x/2);
ax2.scatter(np.random.random(10), np.random.random(10));
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax4.hist(np.random.randn(1000));  # Generate an array of thousand integers
```

## Using subplot with indices

```python
# Subplot option 2 (using indices)
np.random.seed(0)

fig, ax = plt.subplots(nrows = 2, ncols = 2, figsize = (10, 5))

# Plot to each different index
ax[0, 0].plot(x, x/2);
ax[0, 1].scatter(np.random.random(10), np.random.random(10));
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax[1, 1].hist(np.random.randn(1000));  # Generate an array of thousand integers
```

## Pyplot vs Matplotlib oo (Object Oriented) method
* When plotting something quickly, use pyplot
* When plotting something more advanced, use the Object Oriented or OO method

```python
# Object Oriented Method

# Make a new DataFrame from given dataframe of all the people whose age > 50
over_50 = heart_disease[heart_disease["age"] > 50]
over_50.head()
```

```python
# Pyplot method
over_50.plot(kind = "scatter", x = 'age', y = "chol", c = "target")
plt.show()
```

```python
# OO method mixed with pyplot method
fig, ax = plt.subplots(figsize = (10, 6))
over_50.plot(kind = "scatter", x = "age", y = "chol", c = "target", ax = ax)
ax.set_xlim([45, 80])   # Changes x limit from 45 to 80
plt.show()
```

## Object Oriented from scratch

```python
## OO method from scratch
fig, ax = plt.subplots(figsize = (10, 6))

# 1. Plot the data
scatter = ax.scatter(x = over_50["age"], y = over_50["chol"], c = over_50["target"])

# 2. Customize
ax.set(title = "Heart Disease and Cholestrol levels", xlabel = "Age", ylabel = "Cholestrol")

# 3. Add legends
ax.legend(*scatter.legend_elements(), title = "Target")

# 4. Add a horizontal line (i.e. mean of cholestrol)
ax.axhline(over_50["chol"].mean(), linestyle = "--", color = "red")

# 5. Add a vertical line (i.e. mean of age)
ax.axvline(over_50["age"].mode()[0], linestyle = "--", color = "blue")

# 6. Show the plot
plt.show()
```

```python
# Subplot of chol, age, thalach
fig, ax = plt.subplots(nrows = 2, ncols = 1, figsize = (10, 10))

# 1. Add data to ax[0]
scatter = ax[0].scatter(x = over_50["age"], y = over_50["chol"], c = over_50["target"])

# 2. Add a title, xlabel and ylabel
ax[0].set(title = "Heart Disease and Cholestrol levels", xlabel = "Age", ylabel = "Cholestrol")

# 3. Add legend
ax[0].legend(*scatter.legend_elements(), title = "Target")

# 4. Add a chol mean line
ax[0].axhline(over_50["chol"].mean(), linestyle = "--", color = "red")

# -------------------------- Now for ax[1]
# 1. Add data to ax[1]
scatter = ax[1].scatter(x = over_50["age"], y = over_50["thalach"], c = over_50["target"])

# 2. Add a title, xlabel and ylabel
ax[1].set(title = "Heart Disease and Max heart rate", xlabel = "Age", ylabel = "Max Heart Rate")

# 3. Add legend
ax[1].legend(*scatter.legend_elements(), title = "Target")

# 4. Add a chol mean line
ax[1].axhline(over_50["thalach"].mean(), linestyle = "--", color = "red")

# 5. show the plot
plt.show()
```
