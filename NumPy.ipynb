## Dot Product example (nut butter sale)

```python
np.random.seed(0)

sales_amount = np.random.randint(20, size=(5, 3))
sales_amount
```

```python
## Create weekly_sales DataFrame

weekly_sales = pd.DataFrame(sales_amount, index = ["Mon", "Tues", "Wed", "Thurs", "Fri"], columns = ["Almond Butter", "Peanut Butter", "Cashew Butter"])
weekly_sales
```

```python
# Create prices array

prices = np.array([10, 8, 12])
prices, prices.shape
```

```python
# Create butter_prices DataFrame (reshape the prices array because we need one row and three columns)
butter_prices = pd.DataFrame(prices.reshape(1, 3), index = ["Price"], columns = ["Almond Butter", "Peanut Butter", "Cashew Butter"])
butter_prices
```

```python
weekly_sales
```

```python
# Shapes aren't aligned, let's transpose (Normal matrix dot multiplication)

total_sales = prices.dot(sales_amount.T)
total_sales
```

```python
# Matrix dot product with the dataframe
daily_sales = butter_prices.dot(weekly_sales.T)
daily_sales
```

```python
# Daily sales should be of the same shape of weekly sales that's why we use transpose
weekly_sales["Total ($)"] = daily_sales.T
weekly_sales
```
