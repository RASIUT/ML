```python
import nltk

nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(paragraph)

print(sentences)
```

```python
import re

corpus = []
for i in range(len(sentences)):
  review = re.sub('[^a-zA-Z]', ' ', sentences[i])
  review = review.lower()
  corpus.append(review)
```

```python
import re
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

corpus = []
for i in range(len(sentences)):
  review = re.sub('[^a-zA-Z]', ' ', sentences[i])          # Remove anything except a-z and A-Z and instead of that put blank space
  review = review.lower()
  corpus.append(review)


corpus
```

```python
corpus = []

for i in range(len(sentences)):
  review = re.sub('[^a-zA-Z]', ' ', sentences[i])
  review = review.lower()
  review = review.split()

  review = [lemmatizer.lemmatize(word)
            for word in review
            if word not in set(stopwords.words('english'))]

  review = (' ').join(review)

  corpus.append(review)

```

```python
corpus
```

## NGrams

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(binary = True, ngram_range = (3, 3))

X = cv.fit_transform(corpus)
```

## TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize the TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Fit and transform the documents
tfidf_matrix = vectorizer.fit_transform(corpus)

# Get feature names (words)
feature_names = vectorizer.get_feature_names_out()

# Print the TF-IDF scores for each document
for i, doc in enumerate(tfidf_matrix.toarray()):
    print(f"\nDocument {i+1} TF-IDF Scores:")

    # Sort words by TF-IDF score (highest first)
    scores = [(feature_names[j], doc[j]) for j in range(len(feature_names))]        # Creating a list of tuples
    sorted_scores = sorted(scores, key=lambda x: x[1], reverse=True)                # Sorting in descending order

    # Print top scores (non-zero)
    for word, score in sorted_scores:
        if score > 0:
            print(f"  {word}: {score:.4f}")
```
