```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(u"I will google about Meta")
```

```python
doc[2].pos_
```

```python
doc[2].tag_
```

```python
spacy.explain(doc[2].tag_)
```

```python
for word in doc:
  print(word.text, "--->", word.pos_, word.tag_, spacy.explain(word.tag_))
```
