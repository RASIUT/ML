## Feature Engineering Process

### Check for null values

```python
final_df.isnull().sum()
```

### Date_of_Journey is in object(str) format
### So we need to convert it into Date-Time format
### Then we need to pick multiple fields like (Day, Month and Year)

```python
# This code splits the "Date_of_Journey" column in "DD/MM/YYYY" format into day, month, and year
# Using str.split('/'), accessing them via indices [0], [1], and [2].
final_df['Date'] = final_df['Date_of_Journey'].str.split('/').str[0]
final_df['Month'] = final_df['Date_of_Journey'].str.split('/').str[1]
final_df['Year'] = final_df['Date_of_Journey'].str.split('/').str[2]
```

```python
final_df.head(2)
```

```python
# The values of final_df['Date'], final_df['Month'], final_df['Year'] will be all in object(str) so convert them into int
final_df['Date'] = final_df['Date'].astype(int)
final_df['Month'] = final_df['Month'].astype(int)
final_df['Year'] = final_df['Year'].astype(int)
```

```python
final_df.info()
```

```python
# We don't need ['Date_of_Journey'] so we can drop it
final_df.drop('Date_of_Journey', axis = 1, inplace = True)
final_df.head(1)
```

### Arrival_Time & Departure_Time

### Arrival_Time is given in 09:24 10 Jun format
### We don't the date instead we just need the time because we have date for it
### So we'll split it also

```python
final_df['Arrival_Time'] = final_df['Arrival_Time'].str.split(' ').str[0]
final_df.head()
```

```python
# Extract hour and minutes from Arrival_Time
final_df['Arrival_hour'] = final_df['Arrival_Time'].str.split(':').str[0]
final_df['Arrival_min'] = final_df['Arrival_Time'].str.split(':').str[1]
```

```python
final_df.head(2)
```

```python
# Convert the Arrival_hour & Arrival_min into int
final_df['Arrival_hour'] = final_df['Arrival_hour'].astype(int)
final_df['Arrival_min'] = final_df['Arrival_min'].astype(int)
```

```python
final_df.info()
```

```python
# Drop Arrival Time
final_df.drop('Arrival_Time', axis = 1, inplace = True)
```

```python
final_df.head(1)
```

### Departure time

```python
# Split from : which will give [HH, MM] array format use [0] for hrs and [1] for mins
final_df['Dep_hour'] = final_df['Dep_Time'].str.split(':').str[0]
final_df['Dep_min'] = final_df['Dep_Time'].str.split(':').str[1]
```

```python
# Convert datatype form object(str) into int
final_df['Dep_hour'] = final_df['Dep_hour'].astype(int)
final_df['Dep_min'] = final_df['Dep_min'].astype(int)
```

```python
# Drop Dep_time
final_df.drop("Dep_Time", axis = 1, inplace = True)
```

```python
final_df.info()
```

### Route

We don't need Route as we can get to know about total number of stops from **['Total_Stops']** Column and we already have *Source* and *destination*

```python
final_df['Total_Stops'].unique()
```

```python
# There is only one NaN value for Total_Stops so we're gettting that one
final_df[final_df['Total_Stops'].isnull()]
```

```python
# As we get to know the source = "Delhi" and destination = "Cochin" we apply common sense to set Total_Stops = 2
# map the values
final_df['Total_Stops'] = final_df['Total_Stops'].map({'non-stop': 0, '1 stop': 1, '2 stops': 2, '3 stops': 3, '4 stops': 4, 'nan': 2})
```

```python
final_df.head()
```

```python
final_df.drop('Route', axis = 1, inplace = True)
```

```python
final_df.head()
```

### Duration

It is also having object data ( HH h MM m) format

We can convert it to completely minutes

```python
final_df.info()
```

```python
# type() is giving that it is series so we can't apply str.replace() to remove h
type(final_df['Duration'].str.split(' ').str[0])
```

```python
# So for removing h we'll use str.split again
final_df['Duration_hour'] = final_df['Duration'].str.split(' ').str[0].str.split('h').str[0]
final_df.head(2)
```

***There are two rows where *Mumbai to Hyderabad* time is showing as 5m (minutes)***

***This is not possible so we'll delete the rows***

```python
final_df[final_df['Duration_hour'] == '5m']
```

```python
# Drop the two impossible rows using (Index number)
final_df.drop(6474, axis = 0, inplace = True)
final_df.drop(13343, axis = 0, inplace = True)
final_df[final_df['Duration_hour'] == '5m']
```

```python
# Convert duration_hour to int
final_df['Duration_hour'] = final_df['Duration_hour'].astype(int)
final_df.head(2)
```

```python
# Step wise breakdown of
# final_fd['Duration_min'] = final_df['Duration'].str.split(' ').str[1].str.split('m').str[0]

# Get MM m first
final_df['Duration_min'] = final_df['Duration'].str.split(' ').str[1]
```

```python
# Remove m from the values
final_df['Duration_min'] = final_df['Duration_min'].str.split('m').str[0]
```

```python
# Drop the duration
final_df.drop('Duration', axis = 1, inplace = True)
final_df.head(1)
```
