##Import important libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

##Importing the dataset

```python
df = pd.read_csv('/content/Position_Salaries.csv')

X = df.iloc[:,1:-1].values
y = df.iloc[:,-1].values
```

#Training the Random Forest Regressor model on the whole dataset

```python
from sklearn.ensemble import RandomForestRegressor
regressor = RandomForestRegressor(n_estimators = 10, random_state = 0)
regressor.fit(X,y)
```

##Predict a new result

```python
regressor.predict([[6.5]])
```

##Visualise the Random Forest Regressor

```python
from sklearn.tree import plot_tree

fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(20, 15))

# Plot the first 6 trees in the Random Forest
for i, ax in enumerate(axes.flatten()):
    if i < len(regressor.estimators_):
        tree = regressor.estimators_[i]
        plot_tree(tree, ax=ax, filled=True)
        ax.set_title(f'Tree {i+1}')

# Add a suptitle for the overall plot
plt.suptitle('Random Forest Regression')

# Show the plot
plt.show()
```
