##Importing Datasets

```python
df = pd.read_csv("/content/Market_Basket_Optimisation.csv", header = None)
```

##Data Preprocessing

We need list of transactions instead of pandas Data_Frame

```python
transactions = []
for i in range(0, len(df)):
  transactions.append([
        str(df.values[i, j]) for j in range(0, len(df.columns))
      ])
```

##Training the Apriori model on the dataset

```python
from apyori import apriori
rules = apriori(transactions,
                min_support = 0.003,
                min_confidence = 0.2,
                min_lift = 3,
                min_length = 2,
                max_length = 2
               )
```

##Visualizing the results

###Displaying the first results coming directly from the output of the aprior function

```python
results = list(rules)
results
```
