#Multiple Linear Regression

##Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

##Importing Dataset

```python
df = pd.read_csv('/content/50_Startups.csv')
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

**Check for null values**

```python
print(X.isnull().sum())
```

```python
print(y.isnull().sum())
```

```python
X.head()
```

We can see that there is one Categorical column i.e. State so we'll apply One Hot Encoding to create dummy variables


##Encoding Categorical Data

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers = [('encoder', OneHotEncoder(), [3])], remainder = "passthrough")
X = np.array(ct.fit_transform(X))
```

```python
print(X)
```

##Splitting and training the dataset

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
```

##Trainng the Multiple Linear Regression model on the Training set

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

##Predicting the Test set results

```python
y_test = np.array(y_test)
y_pred = regressor.predict(X_test)
np.set_printoptions(precision = 2)

y_pred_vertical = y_pred.reshape(len(y_pred), 1)
y_test_vertical = y_test.reshape(len(y_test), 1)

print(np.concatenate((y_pred_vertical, y_test_vertical), 1))
```

## Evaluating model performance

```python
from sklearn.metrics import r2_score
r2_score(y_test, y_pred)
```

##Making a single prediction (for example the profit of a startup with R&D Spend = 160000, Administration Spend = 130000, Marketing Spend = 300000 and State = 'California')

```python
print(regressor.predict([[1, 0, 0, 160000, 130000, 300000]]))
```

[180892.25]

Therefore, our model predicts that the profit of a Californian startup which spent 160000 in R&D, 130000 in Administration and 300000 in Marketing is $ 181566,92.

Important note 1: Notice that the values of the features were all input in a double pair of square brackets. That's because the "predict" method always expects a 2D array as the format of its inputs. And putting our values into a double pair of square brackets makes the input exactly a 2D array. Simply put:

1,0,0,160000,130000,300000→scalars

[1,0,0,160000,130000,300000]→1D array

[[1,0,0,160000,130000,300000]]→2D array

Important note 2: Notice also that the "California" state was not input as a string in the last column but as "1, 0, 0" in the first three columns. That's because of course the predict method expects the one-hot-encoded values of the state, and as we see in the second row of the matrix of features X, "California" was encoded as "1, 0, 0". And be careful to include these values in the first three columns, not the last three ones, because the dummy variables are always created in the first columns.
